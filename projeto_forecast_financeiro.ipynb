```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
caminho = "/content/drive/MyDrive/Colab Notebooks/projeto_forecast_financeiro/data/raw/ipeadata[24-05-2026-09-39].csv"

df = pd.read_csv(caminho, sep=',')
```

```python
df = df.rename(columns={'Taxa de juros - Selic - fixada pelo Copom - (% a.a.) - Banco Central do Brasil- Boletim- Seção mercado financeiro e de capitais (Bacen/Boletim/M. Finan.) - BM366_TJOVER366': 'Taxa Selic'})
```

```python
print(df.head(3))
```

```python
df = df.drop('Unnamed: 2', axis=1)
```

```python
df['Data'] = pd.to_datetime(df['Data'], dayfirst=True)
```
